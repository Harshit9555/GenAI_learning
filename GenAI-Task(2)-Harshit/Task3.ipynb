```python
total_revenue = 0
orders = []
while True:
    menu = input("Enter the menu:")
    if menu == 'q':
        break
    elif menu == '1':
        try:
            order_amt = int(input("Enter order amount"))
            if order_amt <0:
                print("Can'y be negative")
                continue
            orders.append(order_amt)
            if order_amt >= 2000:
                discount = 0.15
            elif order_amt >= 1500:
                discount = 0.10
            elif order_amt>=1000:
                discount = 0.07
            else:
                discount = 0
            final_amount = order_amt - (order_amt*discount)
            total_revenue += final_amount
            
        except ValueError:
            print("Error: Please Enter valid integer")
    elif menu == '2':
        print(f"All order are: {orders}")
        print(f"Total revenue after discounts: {total_revenue}")
        break
    else:
        print("Invalid menu option.")
    
    
```
