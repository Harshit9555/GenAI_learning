```python
orders = [1200,2500,800,1750,3000]
total_revenue = 0
for order_amount in orders:
    if order_amount >=2000:
        total_revenue = total_revenue + (order_amount- order_amount*0.15)
        print(f"The order amount :{order_amount} applying 15 % discount.The final amount after discount : {order_amount- order_amount*0.15}")
    elif order_amount >=1500 and order_amount <2000:
        total_revenue = total_revenue + (order_amount- order_amount*0.10)
        print(f"The order amount :{order_amount} applying 10 % discount.The final amount after discount : {order_amount- order_amount*0.10}")
    elif order_amount >= 1000 and order_amount < 1500:
        total_revenue = total_revenue + (order_amount- order_amount*0.07)
        print(f"The order amount :{order_amount} applying 7 % discount.The final amount after discount : {order_amount-order_amount*0.07}")
    else:
        print(f"The final amount, No dicount {order_amount}")
print(f"Total renvenue : {total_revenue}")
```
