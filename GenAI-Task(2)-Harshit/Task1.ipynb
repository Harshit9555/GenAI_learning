```python
try:
    order_amount = int(input("Enter the amount :"))

    if order_amount >=2000:
        print(f"The final amount after discount : {order_amount- order_amount*0.15}")
    elif order_amount >=1500 and order_amount <2000:
        print(f"The final amount after discount : {order_amount- order_amount*0.10}")
    elif order_amount >= 1000 and order_amount < 1500:
        print(f"The final amount after discount : {order_amount-order_amount*0.07}")
    else:
        print(f"The final amount, No dicount {order_amount}")
except ValueError:
    print("Error: please enter a valid integer")
```
