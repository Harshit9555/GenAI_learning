```python
price_dict = {
    "Mobile":25000,
    "Pen":50,
    "Bottle":200,
    "Shirt":1500,
    "comb":80.50,
    "football":950
}
```

```python
print(price_dict)
```

```python
price_dict.update({"lamp":350})
price_dict["Pen"] = 70
price_dict.pop("Shirt",None)
print(price_dict)
```

```python
total_sum = 0
n = len(price_dict)
for key,value in price_dict.items():
    total_sum +=value;
print(total_sum/n)
    
    
```

```python
maxi = max(price_dict.values())
mini = min(price_dict.values())
prod = maxi*mini
print(prod)
```
